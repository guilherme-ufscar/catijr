Quartis:Quartis dividem um conjunto de dados ordenados em 4 partes iguais:<br><br>

Q1 (1º quartil) = valor que separa os 25% menores dados<br>
Q2 (2º quartil) = mediana, separa os 50% menores dos 50% maiores<br>
Q3 (3º quartil) = valor que separa os 75% menores dados% finais<br><br>

Dados ordenados do menor ao maior:

<pre>
Dados ordenados do menor ao maior:
|---------|---------|---------|---------|
mín      Q1       Q2       Q3       máx
         25%      50%      75%

</pre>

Quartil Intervalo = Quando não há valores exatos, mas sim uma faixa, ex 160-164, não da para achar q1 exatamente, aí entra o quartil de intervalo.<br><br>

Formula: $$Q_k = L_i + \frac{k \cdot \sum f_i - Fac_{anterior}}{f_{intervalo}} \times h$$

| Símbolo | Significado |
|--------|-------------|
| $L_i$ | Limite inferior da classe do quartil (valor mínimo, ex: 160-164 o LI seria 160) |
| $k$ | 0,25 → Q1 / 0,50 → Q2 / 0,75 → Q3 |
| $\sum f_i$ | Total de dados |
| $Fac_{anterior}$ | Frequência acumulada da classe anterior |
| $f_{intervalo}$ | Frequência da classe do quartil |
| $h$ | Amplitude do intervalo |


### Exemplo — Estatura de 40 alunos:

| Estatura | f | Fac |
|----------|---|-----|
| 160 ├── 164 | 7 | 7  |
| 164 ├── 168 | 4 | 11 |
| 168 ├── 172 | 5 | 16 |
| 172 ├── 176 | 8 | 24 |
| 176 ├── 180 | 16 | 40 |

---

### Calculando Q1 (25%):

**Posição:** $0,25 \times 40 = 10$

A posição 10 está na classe **164 ├── 168** (Fac chega a 11)

$$Q_1 = 164 + \frac{0,25 \cdot 40 - 7}{4} \times 4 = 164 + \frac{10 - 7}{4} \times 4 = 164 + 3 = 167$$

---

### Calculando Q2 (50%):

**Posição:** $0,50 \times 40 = 20$

A posição 20 está na classe **172 ├── 176** (Fac chega a 24)

$$Q_2 = 172 + \frac{0,50 \cdot 40 - 16}{8} \times 4 = 172 + \frac{20 - 16}{8} \times 4 = 172 + 2 = 174$$

---

### Calculando Q3 (75%):

**Posição:** $0,75 \times 40 = 30$

A posição 30 está na classe **176 ├── 180** (Fac chega a 40)

$$Q_3 = 176 + \frac{0,75 \cdot 40 - 24}{16} \times 4 = 176 + \frac{30 - 24}{16} \times 4 = 176 + 1,5 = 177,5$$

Percentil:<br>Serve para dizer aonde está em relação ao grupo

$$P_k = L_i + \frac{\frac{k}{100} \cdot \sum f_i - Fac_{anterior}}{f_{intervalo}} \times h$$

Mesma fórmula do quartil, mas **k** vai de 1 a 100.
<br>Ex: P30 = percentil que separa os 30% menores valores do restante.

| Símbolo | Significado |
|--------|-------------|
| $k$ | Percentil desejado (ex: 30, 75, 90...) |
| $L_i$ | Limite inferior da classe |
| $\sum f_i$ | Total de dados |
| $Fac_{anterior}$ | Frequência acumulada da classe anterior |
| $f_{intervalo}$ | Frequência da classe |
| $h$ | Amplitude do intervalo |

Amplitude interquartil:
<br> Amplitude Interquartil = Qn - Qk

**Boxplot** = representação gráfica boa para ver dispersão dos valores, principalmente os discrepantes<br>
**Outlier** = valor que foge dos demais
