Fatorial:
4! = 4x3x2x1

Permutação Simples (qtd de formas que da para organizar n objetos diferentes) = n!<br>
Arranjo Simples (organizar quando a ordem importa) = (n!)/(n-p)! em que n = qtd de objetos, p = qtd de posições<br>
Combinação Simples (escolher n entre k objetos, ex 3 clientes escolhidos entre 10) (ordem não importa, removemos)= k!/(n!*(k-n!))

Probabilidade:<br>
Espaço amostral (S) = conjunto de todos resultados possíveis, exemplo "moeda = {par,impar}" "dado = {1,2,3,4,5,6}"<br>
Evento (E) = Subconjunto do espaço amostral, exemplo "chance de ocorrer par na moeda" "ocorrer valor impar no dado"<br>

Probabilidade = Evento (Casos Favoraveis)/Espaço amostral (Total de casos), exemplo: valores pares de um dado de 6 valores = 3/6 = 1/2 = 0.5 = 50%

P(A | B) Probabilidade Condicional = Probabilidade do evento A acontecer, dado que o evento B aconteceu.
