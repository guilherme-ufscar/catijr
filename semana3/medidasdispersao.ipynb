Amplitude = maior valor - menor valor<br>
Desvio padrão = Mostra quanto os valores se afastam do centro da média (quanto os valores fogem da média). A média não é tão confiável, porque por exemplo (30 + 300 + 30)/3 = 120, mas um valor "roubou muito" a media

Desvio padrão<br><br>
1. Diferença do valor com a média:<br>
30-120=-90<br>
300-120=180<br>
30-120=-90<br><br>

2. Deixar cada valor em quadrado, então (-90)²=8100, 180²=32400 e (-90)²=8100<br><br>

3. Chegamos na variância, que será a media dos quadrados (8100+32400+8100)/3 = 48600<br><br>

4. Para chegar no desvio padrão faz raiz quadrada da variância, neste caso sqrt(48600) = 220 (aproximadamente)
