```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = sns.load_dataset('titanic')
df.head()
```

```python
qtd_homens = (df['sex'] == 'male').sum()
qtd_mulher = (df['sex'] == 'female').sum()
print(f'Quantidade de homens é: {qtd_homens}\nQuantidade de mulheres: {qtd_mulher}')

homens_survived = ((df['survived'] == 1) & (df['sex'] == 'male')).sum()
mulher_survived = ((df['survived'] == 1) & (df['sex'] == 'female')).sum()
total_passageiros = df['sex'].count() #891 passageiros
#quantidade_nan = df['sex'].isna().sum()
#print(quantidade_nan)
#verificando se tem NaN, não há

print(f'Quantidade de homens sobreviventes: {homens_survived}\nQuantidade de mulheres sobreviventes: {mulher_survived}')


porcentagem_homens_vivo = (homens_survived/total_passageiros)*100
print(f'A porcentagem de homens sobreviventes do total de pessoas é : {porcentagem_homens_vivo:.2f}%')
porcentagem_homens_total_homens = (homens_survived/qtd_homens)*100
print(f'Contudo, a porcentagem de quantidade de homens que sobreviveram, em relação a quantidade total de homens, foi de: {porcentagem_homens_total_homens}')


porcentagem_mulheres_vivo = (mulher_survived/total_passageiros)*100
porcentagem_mulher_total_mulher = (mulher_survived/qtd_mulher)*100
print(f'A porcentagem de mulheres sobreviventes do total de pessoas é : {porcentagem_mulheres_vivo:.2f}%')
print(f'Contudo, a porcentagem de quantidade de mulheres que sobreviveram, em relação a quantidade total de mulheres, foi de: {porcentagem_mulher_total_mulher}')
```

```python
# 1. Verificando se tem NaN nas idades
tem_nan_idade = df['age'].hasnans
print(tem_nan_idade)

media_homens = df.loc[df['sex'] == 'male', 'age'].mean()
media_mulheres = df.loc[df['sex'] == 'female', 'age'].mean()

df.loc[(df['sex'] == 'male') & (df['age'].isna()), 'age'] = media_homens
df.loc[(df['sex'] == 'female') & (df['age'].isna()), 'age'] = media_mulheres
```

```python
idade_min = df['age'].min()
idade_max = df['age'].max()
print(f'A idade da pessoa mais jovem é {idade_min:.0f}, que equivale a {(idade_min*12):.0f} meses')
print(f'A idade da pessoa mais velha é {idade_max:.0f} anos')

print(f'A média da idade dos homens é {media_homens:.0f}')
print(f'A média da idade das mulheres é {media_mulheres:.0f}')

moda_homens = df.loc[df['sex'] == 'male', 'age'].mode()
moda_mulheres = df.loc[df['sex'] == 'female', 'age'].mode()
print(f'As idades que mais aparecem são para mulheres {moda_mulheres[0]:.0f} e para homens {moda_homens[0]:.0f}')
```

```python
# 1. Verificando se tem NaN nas idades
#tem_nan_pclass = df['pclass'].hasnans
#print(tem_nan_pclass)
#total_passageiros
qtd_primeira_classe = (df['pclass'] == 1).sum()
qtd_segunda_classe = (df['pclass'] == 2).sum()
qtd_terceira_classe = (df['pclass'] == 3).sum()

print(f'A quantidade de pessoas na primeira classe é {qtd_primeira_classe}, equivale a {((qtd_primeira_classe/total_passageiros)*100):.0f}% do total')
print(f'A quantidade de pessoas na segunda classe é {qtd_segunda_classe}, equivale a {((qtd_segunda_classe/total_passageiros)*100):.0f}% do total')
print(f'A quantidade de pessoas na terceira classe é {qtd_terceira_classe}, equivale a {((qtd_terceira_classe/total_passageiros)*100):.0f}% do total')

qtd_homem_priclasse = ((df['pclass'] == 1) & (df['sex'] == 'male')).sum()
qtd_homem_segclasse = ((df['pclass'] == 2) & (df['sex'] == 'male')).sum()
qtd_homem_terclasse = ((df['pclass'] == 3) & (df['sex'] == 'male')).sum()
print(f'Existem {qtd_homem_priclasse} homens na primeira classe, {qtd_homem_segclasse} na segunda e {qtd_homem_terclasse} na terceira')
print(f'A quantidade de homens na primeira classe, em relação ao total da primeira classe equivale a {((qtd_homem_priclasse/qtd_primeira_classe)*100):.0f}%, na segunda classe é {((qtd_homem_segclasse/qtd_segunda_classe)*100):.0f}% e na terceira é {((qtd_homem_terclasse/qtd_terceira_classe)*100):.0f}%')

qtd_mulher_priclasse = ((df['pclass'] == 1) & (df['sex'] == 'female')).sum()
qtd_mulher_segclasse = ((df['pclass'] == 2) & (df['sex'] == 'female')).sum()
qtd_mulher_terclasse = ((df['pclass'] == 3) & (df['sex'] == 'female')).sum()
print(f'Existem {qtd_mulher_priclasse} mulheres na primeira classe, {qtd_mulher_segclasse} na segunda e {qtd_mulher_terclasse} na terceira')
print(f'A quantidade de mulheres na primeira classe, em relação ao total da primeira classe equivale a {((qtd_mulher_priclasse/qtd_primeira_classe)*100):.0f}%, na segunda classe é {((qtd_mulher_segclasse/qtd_segunda_classe)*100):.0f}% e na terceira é {((qtd_mulher_terclasse/qtd_terceira_classe)*100):.0f}%')
```
