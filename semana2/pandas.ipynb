<h5>Loc e Iloc</h5>

```python
dados = {
    'produto': ['arroz', 'feijão', 'macarrão'],
    'preço': [10, 8, 5],
    'estoque': [100, 50, 200]
}

indexe = ['a','b','c']

df = pd.DataFrame(dados, index=indexe)

print(df.loc[['a','c'],['produto','estoque']]) #fatiar o que quer mostrar

print(df.iloc[2])
print(df.iloc[0:2,2:3]) #mostra 2 linha (0 a 2) e apenas uma coluna (2:3)
```

<h5>Filtrando Dados</h5>

```python
dados = {
    'nome': ['Guilherme', 'Pedro', 'Ana','José', 'Maria', 'João', 'Carla', 'Fernanda', 'Ricardo', 'Sofia','Guilherme2'],
    'idade': [19,3,10,20, 25, 30, 22, 28, 35, 27,18],
    'estado': ['SP','SP', 'MG', 'PB', 'RJ', 'RS', 'SC', 'BA', 'PE', 'CE','SP'],
    'nota': [100, 5, 33, 85, 43, 93, 21, 41, 77, 88, 99]
}

df = pd.DataFrame(dados)

print(df[df['idade'] < 20])
print()
print(df[(df['estado'] == 'SP') & (df['nota'] > 50)])
```

<h5>Sort Values</h5>

```python
dados = {
    'nome': ['Guilherme', 'Pedro', 'Ana','José', 'Maria', 'João', 'Carla', 'Fernanda', 'Ricardo', 'Sofia','Guilherme2'],
    'idade': [19,3,10,20, 25, 30, 22, 28, 35, 27,18],
    'estado': ['SP','SP', 'MG', 'PB', 'RJ', 'RS', 'SC', 'BA', 'PE', 'CE','SP'],
    'nota': [100, 5, 33, 85, 43, 93, 21, 41, 77, 88, 99]
}

df = pd.DataFrame(dados)

print(df.sort_values(by="nota"))
print(df.sort_values(by="nota", ascending=False)) #decrescenet
```
