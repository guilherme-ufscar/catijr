```python
lista = []

def concluir():
    if len(lista) == 0:
        print('Nenhuma tarefa para concluir.')
        return

    indice = int(input('Selecione o item: '))
    
    if indice < 1 or indice > len(lista):
        print('Item inválido!')
        return

    tarefa = lista[indice - 1] 
    
    for chave in tarefa:
        if tarefa[chave] == True:
            print('Item já concluído!')
        else:
            tarefa[chave] = True 
            print('Tarefa concluída com sucesso!')

def remover():
    if len(lista) == 0:
        print('Nenhuma tarefa para concluir.')
        return

    indice = int(input('Selecione o item: '))
    
    if indice < 1 or indice > len(lista):
        print('Item inválido!')
        return

    tarefa = lista[indice - 1] 
    lista.pop(indice-1)

def visualizar():
    if len(lista) == 0:
        print('Lista vazia')
    else:
        for idx, i in enumerate(lista, start=1):
            for j in i:
                if i[j] == True:
                    print(f'{idx}.[X] {j}')
                else:
                    print(f'{idx}.[ ] {j}')

def adicionar_tarefa():
    desc = input('Adicionar qual tarefa? ')
    lista.append({desc: False})
    print(f'Item adicionado: {lista}')

def menu():
    print("\nMenu:")
    print("1. Adicionar Tarefa")
    print("2. Visualizar Tarefas")
    print("3. Marcar Tarefa como Concluída")
    print("4. Remover Tarefa")
    print("5. Sair")
    
    opcao = ['1', '2', '3', '4', '5']

    escolha = input('Escolha uma opção: ')

    while escolha not in opcao:
        print('Opção inválida!')
        escolha = input('Escolha uma opção: ')

    if escolha == '1':
        print(f'Opção {escolha} selecionada')
        print("1. Adicionar Tarefa")
        adicionar_tarefa()
        menu()

    if escolha == '2':
        print(f'Opção {escolha} selecionada')
        print("2. Visualizar Tarefas")
        visualizar()
        menu()

    if escolha == '3':
        print(f'Opção {escolha} selecionada')
        print("3. Marcar Tarefa como Concluída")
        concluir()
        menu()

    if escolha == '4':
        print(f'Opção {escolha} selecionada')
        print("4. Remover Tarefa")
        remover()
        menu()

    if escolha == '5':
        print(f'Opção {escolha} selecionada')
        print("5. Sair")

menu()
```
