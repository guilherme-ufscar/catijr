```python
import pandas as pd
import sqlite3 as sql
banco = sql.connect(':memory:')

clientes = pd.read_csv('clientes.csv')
compras = pd.read_csv('compras.csv')

clientes.to_sql("clientes",banco,index=False,if_exists='replace')
compras.to_sql("compras",banco,index=False,if_exists='replace')

resultado_inner = pd.read_sql_query("SELECT clientes.cpf, nome, cidade FROM clientes INNER JOIN compras ON clientes.cpf = compras.cpf_cliente", banco)
#mostra só os que tem em comunm

resultado_leftjoin = pd.read_sql_query("SELECT clientes.cpf, nome, cidade FROM clientes left JOIN compras ON clientes.cpf = compras.cpf_cliente", banco)
#todos clientes, mesmo os que não compraram
```
