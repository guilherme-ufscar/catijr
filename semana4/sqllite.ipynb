```python
import sqlite3 as sql
import pandas as pd
```

```python
df = pd.read_csv('clientes.csv')
df.head()
```

```python
banco = sql.connect(':memory:')
df.to_sql("clientes",banco,index=False,if_exists='replace')
```

```python
dfsql = pd.read_sql_query("select * from clientes",banco)
dfsql.head()
```

```python
dffiltro = pd.read_sql_query("select * from clientes where (idade > 30 and idade < 40)",banco)
dffiltro.head()
dffiltro.describe() #tras informacoes gerais
```
