```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
dados_normais = np.random.normal(loc=50, scale=10, size=1000) #1000 valores, com media 50, com desvio em 10 (40 e 60 entao)
dados_com_outliers = np.append(dados_normais, [150, 160, 170]) #aqui adicionamos 3 valores "absurdos", já que a média é 50
normal_data = np.random.normal(loc=0, scale=1, size=1000) #usar quando fica perto da media e tem poucos outliers
long_tail_data = np.random.lognormal(mean=1, sigma=1.5, size=1000) #quando não pode ser negativo,muitos valores baixos e poucos altos,media puxada por causa dos extremos
```

```python
df = pd.DataFrame({'normal': normal_data,'long_tail': long_tail_data})
df.head()
```

```python
plt.hist(dados_normais, bins=30, color='skyblue', edgecolor='black') #dados-normais, 30 barras, cor azul e borda preta
plt.title('Histograma de Distribuição Normal')
plt.xlabel('Valor')
plt.ylabel('Frequência')
plt.show()
```

```python
plt.hist(dados_com_outliers, bins=30, color='skyblue', edgecolor='black')
plt.title('Histograma de Distribuição Normal com outliers')
plt.xlabel('Valor')
plt.ylabel('Frequência')
plt.show()
```

```python
plt.hist(long_tail_data, bins=30, color='skyblue', edgecolor='black')
plt.title('Histograma de Distribuição Normal com outliers')
plt.xlabel('Valor')
plt.ylabel('Frequência')
plt.show()
```
